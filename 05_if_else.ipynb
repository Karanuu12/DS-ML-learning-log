```python
a = int(input("Enter your percentage (Do not use % sign):" ))

if(a <= 35):
    print("You are Fail.")
elif(35<= a <50):
    print("You have got C grade.")
elif(50<= a <65):
    print("You have got B grade.")
elif(65<= a <75):
    print("You have got A grade.")
elif(75<= a <=100):
    print("You have got A+ grade.")
else :
    print("Enter valid marks!")
```
